#### **MESSAGES**


```python
"""
Messages are the fundamental units of context for the llm models. 
They represent the input and output of the models, carring both content and metadata
needed to represent the state of conversation when interacting with an llm
"""
```

```python
import os 
from dotenv import load_dotenv
load_dotenv()

api_key = os.getenv('GROQ_API_KEY')
```

```python
from langchain.chat_models import init_chat_model

groq_llm = init_chat_model(
    model='openai/gpt-oss-120b',
    api_key= api_key,
    model_provider='groq'
    
)
```

```python
groq_llm.invoke('what is llm ?').content
```

Text prompts are strings - ideal for straighforward generation tasks where you don't need to retain coversational history


```python
groq_llm.invoke('what is langchain ?')
```

```python
from langchain.messages import SystemMessage, AIMessage, HumanMessage

messages = [
    SystemMessage('you are an expert in AIML domain'),
    HumanMessage('what is back propogation?')
]

groq_llm.invoke(messages).content
```
