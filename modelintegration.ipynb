```python
import os
from dotenv import load_dotenv
load_dotenv()
groq_api_key = os.getenv('GROQ_API_KEY')
gemini_api_key = os.getenv('GEMINI_API_KEY')
groq_api_key
```

```python
from langchain.chat_models import init_chat_model
llm_model = init_chat_model(
    model='openai/gpt-oss-120b',
    model_provider='groq',
    api_key = groq_api_key
)
llm_model
```

```python
groq_response = llm_model.invoke('How are you buddy ?')
groq_response.content
```

```python
groq_response = llm_model.invoke('what is AI and ML ?')
groq_response.content
```

```python
groq_response = llm_model.stream('what is AI and ML ?')

for chunk in groq_response:
    print(chunk.text, end='|', flush=True)
```

```python
queries = llm_model.batch([
    "What is RAG ?",
    "what is batch norm ?",
    "What is contrastive learning ?"
], config={'max_concurrency' : 5})

for response in queries:
    print(response.text)
```
